````python
import pandas as pd

bench_meta = pd.read_table("DTE-aerial-bench-meta.tab")
bench_meta['id'].unique()
````

````python
train_meta = pd.read_table("DTE-aerial-train-meta.tab")
train_meta['id'].unique()
````

````python
val_meta = pd.read_table("DTE-aerial-val-meta.tab")
val_meta['id'].unique()
````

````python
# Convert IDs to sets
bench_ids = set(bench_meta["id"].astype(str))
train_ids = set(train_meta["id"].astype(str))
val_ids = set(val_meta["id"].astype(str))

# Check overlaps
bench_train_overlap = bench_ids & train_ids
bench_val_overlap = bench_ids & val_ids
train_val_overlap = train_ids & val_ids

print(f"Bench ∩ Train: {len(bench_train_overlap)}")
print(f"Bench ∩ Val: {len(bench_val_overlap)}")
print(f"Train ∩ Val: {len(train_val_overlap)}")

if len(bench_train_overlap) == 0:
    print("✅ No overlap between benchmark and training IDs.")

if len(bench_val_overlap) == 0:
    print("✅ No overlap between benchmark and validation IDs.")

if len(train_val_overlap) == 0:
    print("✅ No overlap between training and validation IDs.")
````
