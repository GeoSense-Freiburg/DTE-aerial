````python
import json

# Load the submitted Croissant metadata, replace with your own file path if needed
with open("DTE-aerial-bench.json", "r") as f:
    croissant = json.load(f)

# ------------------------------------------------------------------
# Add machine-actionable metadata (demonstration only)
# ------------------------------------------------------------------

croissant["dataset_split"] = "test"

croissant["class_encoding"] = [
    {"id": 0, "name": "background"},
    {"id": 1, "name": "tree_cover"},
    {"id": 2, "name": "tree_mortality"}
]

croissant["void_label"] = 255

# ------------------------------------------------------------------
# Save the augmented metadata
# ------------------------------------------------------------------

with open("DTE-aerial-bench-extended.json", "w") as f:
    json.dump(croissant, f, indent=2)

print("Extended Croissant metadata written.")
````
